## 2.1 Derivation: Dephasing Channel

In 2.1, I wrote out an expansion for the density matrix as a function of time

$$
\rho(t) = c_+(t)|\alpha\rangle\langle\alpha| + c_-(t)|-\alpha\rangle\langle-\alpha|
$$

Here I assumed it to be diagonal, but in general there can be off-diagonal terms. I will include a proper derivation below.

I also was sloppy and neglected the buffer space. Technically, I should be writing 

$$\rho_a = \text{Tr}_b(\rho)$$

Let's address the first point. 

Start with the initial state

$$\rho(0) = |\alpha\rangle\langle\alpha|$$

We expand this in the parity basis. There are a couple reasons why the parity basis is nicer to work with. 

- One: the states $|+\rangle, |-\rangle$ are orthogonal, which makes them easier to evaluate quantum channels and traces. Recall these states are even, odd parity states respectively and so they are strictly orthogonal. 
- Two: parity is a symmetry of the Hamiltonian $H_{2ph}$ which only contains $a^2$ terms.

Expanding the initial state in the parity basis:

$$
\rho(0) = \frac{1}{2}|+\rangle\langle +| + \frac{1}{2}|-\rangle\langle -| + \frac{1}{2}|+\rangle\langle -| + \frac{1}{2}|-\rangle\langle +|
$$

Note here I am assuming normalization $\langle \alpha|-\alpha\rangle \approx 0$. Notice that the initial state has off-diagonal terms. This state is still normalized though because when we take the trace the off-diagonal terms zero out from orthogonality.

Over-time as bit-flip errors accumulate, the off-diagonals will decohere:

$$
\rho(t) = \frac{1}{2}|+\rangle\langle +| + \frac{1}{2}|-\rangle\langle -| + \frac{e^{-\gamma t}}{2}|+\rangle\langle -| + \frac{e^{-\gamma t}}{2}|-\rangle\langle +|
$$

Notice this is exactly the dephasing channel $\mathcal{E}_{dephase}(\rho)$!

To obtain the above expression, we see how a bit-flip operator $X_L$ that forces $\vert{}\alpha\rangle \leftrightarrow \vert{}-\alpha\rangle$ acts on the parity basis:

$$X_L \vert{}\mathcal{C}_\alpha^+\rangle = X_L \left[ \frac{\vert{}\alpha\rangle + \vert{}-\alpha\rangle}{\sqrt{2}} \right] = \frac{\vert{}-\alpha\rangle + \vert{}\alpha\rangle}{\sqrt{2}} = \vert{}\mathcal{C}_\alpha^+\rangle$$

$$X_L \vert{}\mathcal{C}_\alpha^-\rangle = X_L \left[ \frac{\vert{}\alpha\rangle - \vert{}-\alpha\rangle}{\sqrt{2}} \right] = \frac{\vert{}-\alpha\rangle - \vert{}\alpha\rangle}{\sqrt{2}} = -\vert{}\mathcal{C}_\alpha^-\rangle$$


We see that a bit-flip acts as a phase-flip in the parity basis. This is why when we expand $\rho(t)$ we should expect a dephasing channel.

Now taking the $t \rightarrow \infty$ limit, we arrive at

$$
\rho(t \rightarrow \infty) = \frac{1}{2}|+\rangle\langle +| + \frac{1}{2}|-\rangle\langle -| = \frac{1}{2}|\alpha\rangle\langle \alpha| + \frac{1}{2}|-\alpha\rangle\langle -\alpha|
$$

This is the maximally mixed state whose $\langle a\rangle = 0$.

So strictly speaking, my derivation in the submission was incomplete. But I arrived at the correct final expression for $\rho(t \rightarrow \infty)$.

## Two-photon Stabilization

In task 1, we saw that in the effective single-mode picture $|\pm\alpha\rangle$ are stable dark states because they are eigenstates of the jump operator $a^2 - \alpha^2$.

But as I showed in 2.1, this single-mode approximation is not valid in task 2. We thus have to work with two separete jump operators $a, b$.

But we've also seen in task 1 that when the system is subject to single-photon dissipation $a$, it is driven from $|\alpha\rangle \rightarrow |0\rangle$. So what is preventing this process from occurring? 

Recall the two-mode Hamiltonian we are working with 

$$\hat H_{2\mathrm{ph}} = g_2\hat a^2 \hat b^\dagger + g_2\hat a^{\dagger 2}\hat b - \epsilon_d\hat b - \epsilon_d\hat b^\dagger$$

Let's look at how the buffer mode evolves in time under this Hamiltonian and the dissipation/jump term $\sqrt{\kappa_b} b$:

$$\dot{b} = -i[b, H] - \frac{\kappa_b}{2}b$$

This is the Heisenberg-Langevin equation. One can derive it by taking the time-derivative of $\langle b\rangle$

$$\frac{d}{dt}\langle b \rangle = \text{Tr}(b\dot{\rho})$$

then plugging in the Lindblad equation for $$\dot{\rho} with the jump operator $\sqrt{\kappa_b}b$.

Back to our equation of motion. If we evaluate the commutator we get

$$\dot{b} = -i(g_2 a^2 - \epsilon_d) - \frac{\kappa_b}{2}b$$

Because the buffer is highly dissipative ($\kappa_b$), it constantly tries to reach a steady state where $\dot{b} = 0$. Setting $\dot{b} = 0$ we get 

$$b = \frac{-i(g_2 a^2 - \epsilon_d)}{\kappa_b/2}$$

If we factor out $g_2$ and define a new constant $\alpha^2 \equiv \epsilon_d / g_2$, this becomes:

$$b \propto (a^2 - \alpha^2)$$

There are three cases.

- Too many photons $a^2 > \alpha^2$: If the memory mode has an amplitude larger than the target $\alpha$, the $g_2 a^2 b^\dagger$ term dominates the drive. Pairs of photons are removed and the buffer photons immediately leak into the environmentvia $\kappa_b$. Energy is drained from the memory mode until it shrinks back to $\alpha$.
- Too few photons $a^2 < \alpha^2$: The drive term $-\epsilon_d(b + b^\dagger)$ dominates. The classical drive pumps photons into the buffer. Energy is pumped from the drive, through the buffer, and into the memory mode, growing its amplitude up to $\alpha$.
- Just right amount of photons  $a^2 = \alpha^2$: This is the dark state

We thus see that this dissipative buffer is what stabilizes the memory!


## 2.2: Another Dephasing Channel

The derivation for 2.2 is identical in structure to 2.1.

Start off with an initial pure state in the parity basis

$$\rho(0) = \vert{}+\rangle\langle+\vert{}$$

Expand it in the coherent basis 

$$\rho(0) = \frac{1}{2}\vert{}\alpha\rangle\langle\alpha\vert{} + \frac{1}{2}\vert{}-\alpha\rangle\langle-\alpha\vert{} + \frac{1}{2}\vert{}\alpha\rangle\langle-\alpha\vert{} + \frac{1}{2}\vert{}-\alpha\rangle\langle\alpha\vert{}$$

The dominant error now is caused by the jump operator $a$:

$$Z\vert{}\pm\rangle = a\vert{}\pm\rangle = \alpha\vert{}\mp\rangle$$

These manifest as bit-flips in the parity basis.

In the coherent basis, these manifest as phase-flips

$$Z\vert{}\pm\alpha\rangle = a\vert{}\pm\alpha\rangle = \pm\alpha\vert{}\alpha\rangle$$

Every time a photon leaks into the environment, the environment "measures" which coherent state the system is in. This continuous measurement extracts information, and in quantum mechanics, information extraction destroys superpositions.

Because single-photon loss does not transfer population between $\vert{}\alpha\rangle$ and $\vert{}-\alpha\rangle$ (it leaves the diagonals alone), its only effect is to kill the off-diagonal coherence terms at a rate proportional to how distinguishable the states are ($\gamma_\phi \propto \kappa_a \vert{}\alpha\vert{}^2$).

Thus, we get the same phase damping channel

$$\rho(t) = \frac{1}{2}\vert{}\alpha\rangle\langle\alpha\vert{} + \frac{1}{2}\vert{}-\alpha\rangle\langle-\alpha\vert{} + \frac{e^{-\gamma_\phi t}}{2}\vert{}\alpha\rangle\langle-\alpha\vert{} + \frac{e^{-\gamma_\phi t}}{2}\vert{}-\alpha\rangle\langle\alpha\vert{}$$

As $t \to \infty$, the off-diagonal phase coherence is completely wiped out by the environment. 

$$\rho(t \to \infty) = \frac{1}{2}\vert{}\alpha\rangle\langle\alpha\vert{} + \frac{1}{2}\vert{}-\alpha\rangle\langle-\alpha\vert{} = \frac{1}{2}|+\rangle\langle +| + \frac{1}{2}|-\rangle\langle -|$$

This is the exact final expression I derived in 2.2.
