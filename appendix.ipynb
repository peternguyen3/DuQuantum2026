## 2.1 Derivation: Dephasing Channel

In 2.1, I wrote out an expansion for the density matrix as a function of time

$$
\rho(t) = c_+(t)|\alpha\rangle\langle\alpha| + c_-(t)|-\alpha\rangle\langle-\alpha|
$$

Here I assumed it to be diagonal, but in general there can be off-diagonal terms. I will include a proper derivation below.

I also was sloppy and neglected the buffer space. Technically, I should be writing 

$$\rho_a = \text{Tr}_b(\rho)$$

Let's address the first point. 

Start with the initial state

$$\rho(0) = |\alpha\rangle\langle\alpha|$$

We expand this in the parity basis. There are a couple reasons why the parity basis is nicer to work with. 

- One: the states $|+\rangle, |-\rangle$ are orthogonal, which makes them easier to evaluate quantum channels and traces. Recall these states are even, odd parity states respectively and so they are strictly orthogonal. 
- Two: parity is a symmetry of the Hamiltonian $H_{2ph}$ which only contains $a^2$ terms.

Expanding the initial state in the parity basis:

$$
\rho(0) = \frac{1}{2}|+\rangle\langle +| + \frac{1}{2}|-\rangle\langle -| + \frac{1}{2}|+\rangle\langle -| + \frac{1}{2}|-\rangle\langle +|
$$

Note here I am assuming normalization $\langle \alpha|-\alpha\rangle \approx 0$. Notice that the initial state has off-diagonal terms. This state is still normalized though because when we take the trace the off-diagonal terms zero out from orthogonality.

Over-time as bit-flip errors accumulate, the off-diagonals will decohere:

$$
\rho(t) = \frac{1}{2}|+\rangle\langle +| + \frac{1}{2}|-\rangle\langle -| + \frac{e^{-\gamma t}}{2}|+\rangle\langle -| + \frac{e^{-\gamma t}}{2}|-\rangle\langle +|
$$

Notice this is exactly the dephasing channel $\mathcal{E}_{dephase}(\rho)$!

To obtain the above expression, we see how a bit-flip operator $X_L$ that forces $\vert{}\alpha\rangle \leftrightarrow \vert{}-\alpha\rangle$ acts on the parity basis:

$$X_L \vert{}\mathcal{C}_\alpha^+\rangle = X_L \left[ \frac{\vert{}\alpha\rangle + \vert{}-\alpha\rangle}{\sqrt{2}} \right] = \frac{\vert{}-\alpha\rangle + \vert{}\alpha\rangle}{\sqrt{2}} = \vert{}\mathcal{C}_\alpha^+\rangle$$

$$X_L \vert{}\mathcal{C}_\alpha^-\rangle = X_L \left[ \frac{\vert{}\alpha\rangle - \vert{}-\alpha\rangle}{\sqrt{2}} \right] = \frac{\vert{}-\alpha\rangle - \vert{}\alpha\rangle}{\sqrt{2}} = -\vert{}\mathcal{C}_\alpha^-\rangle$$


We see that a bit-flip acts as a phase-flip in the parity basis. This is why when we expand $\rho(t)$ we should expect a dephasing channel.

Now taking the $t \rightarrow \infty$ limit, we arrive at

$$
\rho(t \rightarrow \infty) = \frac{1}{2}|+\rangle\langle +| + \frac{1}{2}|-\rangle\langle -| = \frac{1}{2}|\alpha\rangle\langle \alpha| + \frac{1}{2}|-\alpha\rangle\langle -\alpha|
$$

This is the maximally mixed state whose $\langle a\rangle = 0$.

So strictly speaking, my derivation in the submission was incomplete. But I arrived at the correct final expression for $\rho(t \rightarrow \infty)$.
